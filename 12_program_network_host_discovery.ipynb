# Program 12: Network Host Discovery – Lab Network

## Aim:
Write a Python program to check a predefined range of IP addresses in an authorized lab network and display the hosts that respond.

## Concepts Covered:
* Network mapping and active node identification using low-level socket connections
* Multi-threaded asynchronous port scanning using `concurrent.futures`
* Port-based reachability validation checks (TCP handshake targeting standard port 80/443/etc.)


```python
import socket
from concurrent.futures import ThreadPoolExecutor

def probe_host(ip, port=80, timeout=0.2):
    """Attempts a raw TCP handshake on a specific port to check if a host is active."""
    try:
        # Create a standard TCP Socket stream channel
        with socket.socket(socket.AF_INET, socket.SOCK_STREAM) as sock:
            sock.settimeout(timeout)
            # Attempt to establish an active socket connection boundary
            result = sock.connect_ex((ip, port))
            
            # If connect_ex returns 0, the handshake was successful (host is alive)
            if result == 0:
                return (ip, True)
    except Exception:
        pass
    return (ip, False)

def run_host_discovery():
    print("=== NETWORK HOST DISCOVERY UTILITY ===")
    print("[Authorized Lab Environment Mode]\n")
    
    # Define a default local loopback subnet range for safe, fast local testing
    # You can change this base string to any target network subnet configuration
    subnet_base = "127.0.0." 
    start_octet = 1
    end_octet = 15  # Scans 127.0.0.1 through 127.0.0.10
    
    # Generate list of target IP strings
    target_ips = [f"{subnet_base}{i}" for i in range(start_octet, end_octet + 1)]
    
    print(f"[Processing] Scanning network range: {subnet_base}{start_octet} to {subnet_base}{end_octet}")
    print(f"Probing target hosts via active TCP handshakes... Please wait...\n")
    
    active_hosts = []
    
    # Use a ThreadPoolExecutor to run the scans in parallel so the cell finishes instantly
    with ThreadPoolExecutor(max_workers=10) as executor:
        # Submit mapping lookup operations across background worker threads
        results = executor.map(lambda ip: probe_host(ip, port=135, timeout=0.1), target_ips)
        
        for ip, is_alive in results:
            if is_alive:
                active_hosts.append(ip)
                
    # Display the final network discovery logs
    print("=" * 45)
    print(" DISCOVERY RESOLUTION DASHBOARD")
    print("=" * 45)
    if active_hosts:
        print(f"Total Active Hosts Discovered: {len(active_hosts)}")
        print("-" * 45)
        for host in active_hosts:
            print(f" -> [ONLINE] Host Node: {host}")
    else:
        print("No active matching hosts responded within the timeout window.")
    print("=" * 45 + "\n")

if __name__ == "__main__":
    run_host_discovery()
                                                                                                                                                                                            
```
