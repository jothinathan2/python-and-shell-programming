# Program 17: HTTP Request Using Socket – Web Client

## Aim:
Write a Python program using the socket module to connect to a permitted HTTP test server and send a basic GET request to retrieve a webpage.

## Concepts Covered:
* Manual HTTP text protocol construction over standard TCP byte streams
* Transport Layer connection rules (`socket.SOCK_STREAM`)
* Parsing and reading chunks from raw server streaming buffers


```python
import socket

def send_raw_http_get():
    print("=== HTTP REQUEST VIA RAW SOCKET ===")
    
    # Define a reliable, open HTTP test mirror host (safe to probe via raw sockets)
    host = "example.com"
    port = 80  # Standard HTTP port
    
    print(f"[Processing] Resolving domain name and opening socket to {host}:{port}...")
    
    try:
        # 1. Create a low-level standard TCP socket
        with socket.socket(socket.AF_INET, socket.SOCK_STREAM) as client_socket:
            # Set a timeout threshold so the operation doesn't hang indefinitely
            client_socket.settimeout(5.0)
            
            # 2. Establish connection boundary with the remote web host server
            client_socket.connect((host, port))
            print(f"[Connected] Network pipeline open. Sending raw text headers...")
            
            # 3. Construct raw HTTP GET Request string manually
            # Critical formatting: HTTP headers MUST end with a double carriage return line break (\r\n\r\n)
            http_request = (
                "GET / HTTP/1.1\r\n"
                f"Host: {host}\r\n"
                "User-Agent: RawSocketClient/1.0\r\n"
                "Connection: close\r\n"
                "\r\n"
            )
            
            # Encode string layout into bytes and transmit over network buffer channel
            client_socket.sendall(http_request.encode('utf-8'))
            print("[Sent] Payload stream written successfully. Awaiting data response stream...\n")
            
            # 4. Receive data packet loops sequentially from the response stream
            raw_response = b""
            while True:
                buffer_chunk = client_socket.recv(4096)  # Read in chunks of 4096 bytes
                if not buffer_chunk:
                    break  # Stream closed by the host side
                raw_response += buffer_chunk
                
        # 5. Decode the composite byte matrix payload into human-readable text
        decoded_output = raw_response.decode('utf-8', errors='ignore')
        
        # Split header metadata apart from the body for clean visual parsing
        print("=" * 65)
        print(" WEB SERVER STREAM RESPONSE DATA (RAW HEADER + HTML snippet)")
        print("=" * 65)
        
        # Display only first 25 lines of output to avoid excessively cluttering the Notebook UI
        lines = decoded_output.split('\n')
        for line in lines[:25]:
            print(line.rstrip())
            
        if len(lines) > 25:
            print("\n... [Output truncated here to maximize Notebook readability] ...")
        print("=" * 65 + "\n")
        
    except socket.gaierror:
        print("[Network Error] Could not resolve host name domain server.")
    except socket.timeout:
        print("[Timeout Error] Connection or stream response reading exceeded safety window limit.")
    except Exception as e:
        print(f"[Unexpected Error] Communication layer faulted: {e}")

if __name__ == "__main__":
    send_raw_http_get()
```
