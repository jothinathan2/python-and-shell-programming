# Program 6: Service Availability Checker – Socket

## Aim:
Write a Python program to check whether a specified port is open on a localhost or authorized test server using sockets and display the result.

## Concepts Covered:
* Port scanning and network probe operations using sockets
* Setting connection timeout ceilings with `socket.settimeout()`
* Low-level connection handling using `socket.connect_ex()` (returns error codes instead of throwing exceptions)


```python
import socket

def check_service_availability():
    print("=== SERVICE AVAILABILITY CHECKER ===")
    
    # Prompt user for target server host configuration
    host = input("Enter target host (Press Enter for localhost / 127.0.0.1): ").strip()
    if not host:
        host = '127.0.0.1'
        
    # Prompt user for the destination port number to scan
    try:
        port = int(input("Enter port number to verify (e.g., 80, 443, 65432): ").strip())
    except ValueError:
        print("\n[Input Error] Port number must be a valid integer.")
        return

    print(f"\n[Processing] Probing service availability on {host}:{port}...")
    
    try:
        # Create a standard TCP Socket
        with socket.socket(socket.AF_INET, socket.SOCK_STREAM) as sock:
            # Set a timeout threshold so the scan doesn't hang indefinitely
            sock.settimeout(2.0)
            
            # connect_ex returns 0 if the connection succeeded (port open), 
            # otherwise it returns a specific error code integer (port closed/filtered)
            result = sock.connect_ex((host, port))
            
            print("\n" + "="*45)
            if result == 0:
                print(f"[STATUS] PORT {port} IS OPEN!")
                print(f"Service status: Available / Active on '{host}'")
            else:
                print(f"[STATUS] PORT {port} IS CLOSED.")
                print(f"Service status: Unavailable / Closed (Error Code: {result})")
            print("="*45 + "\n")
            
    except socket.gaierror:
        print(f"\n[Network Error] Could not resolve host profile for '{host}'.")
    except Exception as e:
        print(f"\n[Unexpected Error] Failed to complete network check: {e}")

if __name__ == "__main__":
    check_service_availability()
```
