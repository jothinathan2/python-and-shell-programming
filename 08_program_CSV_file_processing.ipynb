# Program 8: CSV File Processing – Employee Records

## Aim:
Write a Python program to read employee records from a CSV file and display employees whose salary is above a specified amount. Handle file-related exceptions.

## Concepts Covered:
* Parsing delimited tabular data using Python's standard `csv` engine
* Numeric data casting and filtering conditions inside row loops
* File error containment via structural `try-except` filters


```python
import csv
import os

def create_sample_csv(filename):
    """Generates a dummy CSV file to ensure the program runs instantly."""
    sample_data = [
        ["EmpID", "Name", "Department", "Salary"],
        ["101", "Alice Smith", "Engineering", "85000"],
        ["102", "Bob Jones", "Marketing", "45000"],
        ["103", "Charlie Brown", "Finance", "95000"],
        ["104", "Diana Prince", "Engineering", "120000"],
        ["105", "Evan Wright", "HR", "50000"]
    ]
    try:
        with open(filename, mode='w', newline='', encoding='utf-8') as file:
            writer = csv.writer(file)
            writer.writerows(sample_data)
        print(f"[System Init] Sample file '{sample-utf8}' verified/generated successfully.\n")
    except Exception as e:
        print(f"[System Init Error] Could not generate sample file: {e}\n")

def filter_employees_by_salary(filename):
    print("=== EMPLOYEE RECORDS FILTER ===")
    
    # Prompt user for the target salary threshold
    try:
        threshold = float(input("Enter minimum salary threshold to filter by: ").strip())
    except ValueError:
        print("\n[Input Error] Please enter a valid numeric value for the salary threshold.\n")
        return

    print(f"\n[Processing] Searching for employees earning more than ${threshold:,.2f}...\n")
    
    try:
        # Attempt to open and parse the data stream
        with open(filename, mode='r', newline='', encoding='utf-8') as file:
            # DictReader mapping headers directly to row values
            csv_reader = csv.DictReader(file)
            
            # Print tabular layout header elements
            print("=" * 60)
            print(f"{'Emp ID':<10}{'Employee Name':<20}{'Department':<15}{'Salary ($)':<15}")
            print("=" * 60)
            
            match_count = 0
            for row in csv_reader:
                try:
                    # Extract and transform row-level payload strings to numeric data
                    salary = float(row['Salary'])
                    
                    if salary > threshold:
                        print(f"{row['EmpID']:<10}{row['Name']:<20}{row['Department']:<15}{salary:<15,.2f}")
                        match_count += 1
                except (ValueError, KeyError):
                    # Gracefully skip rows with corrupt structures or broken columns
                    continue
            
            print("=" * 60)
            print(f"Total Matches Found: {match_count}")
            print("=" * 60 + "\n")
            
    except FileNotFoundError:
        print(f"[Exception Error] The source file '{filename}' was not found.")
        print("Please check the absolute pathway directory or run the file setup steps again.")
    except PermissionError:
        print(f"[Exception Error] Read permissions denied for '{filename}'. Close the spreadsheet if open.")
    except Exception as e:
        print(f"[Unexpected Error] Structural failure during streaming operation: {e}")

if __name__ == "__main__":
    CSV_FILE_NAME = "employee_records.csv"
    
    # Pre-populate sample test registry
    create_sample_csv(CSV_FILE_NAME)
    
    # Fire up calculation query dashboard
    filter_employees_by_salary(CSV_FILE_NAME)
```
