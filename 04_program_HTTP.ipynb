# Program 4: Website Status Checker – HTTP

## Aim:
Write a Python program to send an HTTP request to a permitted website and display its status code and response time.

## Concepts Covered:
* Network communication using Python's standard `urllib.request` library
* Execution profiling via the `time` module to track response latency
* Structural validation using HTTP standard status return flags


```python
import urllib.request
import urllib.error
import time
import socket  # Added missing import to fix the execution error

def check_website_status():
    print("=== WEBSITE STATUS CHECKER ===")
    
    # Prompt user for input target URL
    url = input("Enter the website URL to test (e.g., https://google.com): ").strip()
    
    # Simple formatting validation check to ensure protocol schema is included
    if not url.startswith("http://") and not url.startswith("https://"):
        print("[System Info] Prefixing URL with 'https://' schema...")
        url = "https://" + url

    print(f"\n[Processing] Sending GET request packet to: {url}")
    
    try:
        # Mark initial timestamp before request execution
        start_time = time.time()
        
        # Inject standard user-agent string headers to avoid server blocking filters
        req = urllib.request.Request(
            url, 
            headers={'User-Agent': 'Mozilla/5.0 (Windows NT 10.0; Win64; x64)'}
        )
        
        # Dispatch the connection channel (Timeout ceiling set at 10 seconds)
        with urllib.request.urlopen(req, timeout=10) as response:
            status_code = response.getcode()
            
            # Compute total elapsed delta time
            end_time = time.time()
            elapsed_time = (end_time - start_time) * 1000  # Convert to milliseconds
            
            print("\n" + "="*45)
            print(f"[SUCCESS] Connection Established Successfully!")
            print(f"Target URL:    {url}")
            print(f"Status Code:   {status_code} (OK)")
            print(f"Response Time: {elapsed_time:.2f} ms")
            print("="*45 + "\n")
            
    except urllib.error.HTTPError as e:
        # The server responded, but with an error status code (e.g., 404, 403, 500)
        end_time = time.time()
        elapsed_time = (end_time - start_time) * 1000
        print("\n" + "="*45)
        print(f"[HTTP Error] Server returned a fallback error code.")
        print(f"Status Code:   {e.code} ({e.reason})")
        print(f"Response Time: {elapsed_time:.2f} ms")
        print("="*45 + "\n")
        
    except urllib.error.URLError as e:
        # The server could not be reached at all (e.g., bad domain, down network)
        print("\n" + "="*45)
        print(f"[Network Error] Target server connection failed.")
        print(f"Reason:        {e.reason}")
        print("="*45 + "\n")
        
    except socket.timeout:
        print(f"\n[Timeout Error] Request to '{url}' exceeded the 10-second threshold.")
    except Exception as e:
        print(f"\n[Unexpected Error] An issue occurred: {e}")

if __name__ == "__main__":
    check_website_status()
```
