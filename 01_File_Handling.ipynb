# Program 1: Student Attendance – File Handling

## Aim:
Write a Python program for a college to store student attendance details in a file and read the attendance records. Handle exceptions if the file does not exist.

## Concepts Covered:
* File I/O operations (`w` mode for writing, `r` mode for reading)
* Context managers (`with` statement)
* Exception handling (`try-except` blocks for `FileNotFoundError`)


```python
import os

def write_attendance(filename):
    """Writes student attendance records to a file."""
    print("--- Record Student Attendance ---")
    try:
        with open(filename, 'w') as file:
            # Writing headers to the file
            file.write("Roll No,Name,Status\n")
            
            while True:
                roll_no = input("Enter Roll Number (or 'exit' to stop): ").strip()
                if roll_no.lower() == 'exit':
                    break
                name = input("Enter Student Name: ").strip()
                status = input("Enter Attendance Status (Present/Absent): ").strip()
                
                # Write formatted string to file
                file.write(f"{roll_no},{name},{status}\n")
            print(f"\n[Success] Attendance records successfully saved to '{filename}'.\n")
    except Exception as e:
        print(f"[Error] An error occurred while writing to file: {e}")

def read_attendance(filename):
    """Reads and displays student attendance records from a file with exception handling."""
    print("--- Read Student Attendance Records ---")
    try:
        with open(filename, 'r') as file:
            records = file.readlines()
            
            if len(records) <= 1:
                print("The attendance file is empty.")
                return
                
            print("\n" + "="*40)
            print(f"{'Roll No':<10}{'Student Name':<20}{'Status':<10}")
            print("="*40)
            
            # Skip the first line (header) and read the records
            for line in records[1:]:
                roll_no, name, status = line.strip().split(',')
                print(f"{roll_no:<10}{name:<20}{status:<10}")
            print("="*40 + "\n")
            
    except FileNotFoundError:
        print(f"[Exception Error] The file '{filename}' does not exist yet. Please record attendance first.")
    except Exception as e:
        print(f"[Error] An unexpected error occurred: {e}")

# Main execution loop
if __name__ == "__main__":
    FILE_NAME = "student_attendance.csv"
    
    while True:
        print("=== COLLEGE ATTENDANCE SYSTEM ===")
        print("1. Record Attendance (Write to File)")
        print("2. Display Attendance Records (Read from File)")
        print("3. Exit")
        
        choice = input("Enter your choice (1/2/3): ").strip()
        print()
        
        if choice == '1':
            write_attendance(FILE_NAME)
        elif choice == '2':
            read_attendance(FILE_NAME)
        elif choice == '3':
            print("Exiting system. Goodbye!")
            break
        else:
            print("Invalid choice! Please select 1, 2, or 3.\n")
```
