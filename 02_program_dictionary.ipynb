# Program 2: Product Inventory – Dictionary

## Aim:
Write a Python program to maintain product details using a dictionary. Store product ID, name, quantity and price, and display the available inventory.

## Concepts Covered:
* Python Dictionaries (`dict`) for structured data storage
* Nested data structures (Dictionary of Dictionaries)
* Formatted table output string manipulation


```python
def add_product(inventory):
    """Adds a new product or updates an existing product in the inventory dictionary."""
    print("--- Add / Update Product ---")
    prod_id = input("Enter Product ID: ").strip()
    name = input("Enter Product Name: ").strip()
    
    try:
        quantity = int(input("Enter Quantity: "))
        price = float(input("Enter Price per Unit: "))
        
        # Storing details using Product ID as the unique key
        inventory[prod_id] = {
            "name": name,
            "quantity": quantity,
            "price": price
        }
        print(f"\n[Success] Product '{name}' added/updated successfully.\n")
    except ValueError:
        print("\n[Error] Invalid input for quantity or price. Please enter numeric values.\n")

def display_inventory(inventory):
    """Displays the entire available inventory in a clean tabular format."""
    print("--- Current Product Inventory ---")
    if not inventory:
        print("The inventory is currently empty.\n")
        return
        
    print("\n" + "="*55)
    print(f"{'Prod ID':<10}{'Product Name':<20}{'Quantity':<12}{'Price ($)':<10}")
    print("="*55)
    
    # Iterate through the dictionary to display details
    for prod_id, details in inventory.items():
        print(f"{prod_id:<10}{details['name']:<20}{details['quantity']:<12}{details['price']:<10.2f}")
    print("="*55 + "\n")

# Main execution loop
if __name__ == "__main__":
    # Primary dictionary storing the inventory database
    product_inventory = {}
    
    while True:
        print("=== INVENTORY MANAGEMENT SYSTEM ===")
        print("1. Add/Update Product")
        print("2. Display Available Inventory")
        print("3. Exit")
        
        choice = input("Enter your choice (1/2/3): ").strip()
        print()
        
        if choice == '1':
            add_product(product_inventory)
        elif choice == '2':
            display_inventory(product_inventory)
        elif choice == '3':
            print("Exiting inventory system. Goodbye!")
            break
        else:
            print("Invalid choice! Please select 1, 2, or 3.\n")
```
