# Program 18: Exception Handling – ATM System

## Aim:
Write a Python program to simulate an ATM system that handles invalid PIN input, insufficient balance, invalid withdrawal amount and other runtime errors.

## Concepts Covered:
* Object-Oriented Exception Frameworks (Custom user-defined exceptions)
* Structural state verification checks and transaction boundaries
* Graceful runtime error recovery inside nested interactive loops


```python
# =====================================================================
# Custom User-Defined Exception Architecture
# =====================================================================
class InvalidPINError(Exception):
    """Raised when user credential checks fail match validation."""
    pass

class InsufficientBalanceError(Exception):
    """Raised when target debit volumes exceed current account equity."""
    pass

class InvalidWithdrawalAmountError(Exception):
    """Raised when requested transaction units break allocation increments (e.g. not multiples of 10 or 100)."""
    pass

# =====================================================================
# ATM System Simulator Engine
# =====================================================================
class ATMSimulator:
    def __init__(self, initial_balance=5000.0, true_pin="1234"):
        self.balance = initial_balance
        self.pin = true_pin
        self.pin_attempts = 0

    def authenticate_user(self):
        """Validates incoming interface user credentials."""
        entered_pin = input("Enter your 4-digit secret ATM PIN: ").strip()
        if entered_pin != self.pin:
            self.pin_attempts += 1
            remaining = 3 - self.pin_attempts
            if self.pin_attempts >= 3:
                raise InvalidPINError("CRITICAL: Account locked due to 3 consecutive incorrect PIN entries.")
            else:
                raise InvalidPINError(f"ACCESS DENIED: Incorrect PIN code. Remaining attempts: {remaining}")
        
        # Reset counter on successful verification entry
        self.pin_attempts = 0
        print("[SUCCESS] PIN Authenticated successfully. Opening banking dashboard...")

    def withdraw_cash(self):
        """Executes banking ledger debits under complex validation logic checks."""
        try:
            amount = float(input("Enter cash withdrawal amount ($): ").strip())
        except ValueError:
            raise InvalidWithdrawalAmountError("Input format error! Transaction requires valid numeric quantities.")

        # Validation Rule 1: Zero or Negative Values
        if amount <= 0:
            raise InvalidWithdrawalAmountError("Transaction blocked: Amount must be greater than zero.")

        # Validation Rule 2: Increment Checks (e.g., standard bills configuration matching)
        if amount % 100 != 0 and amount % 10 != 0:
            raise InvalidWithdrawalAmountError("Transaction blocked: ATM only dispenses cash notes of $10 or $100.")

        # Validation Rule 3: Balance Threshold Checks
        if amount > self.balance:
            raise InsufficientBalanceError(f"Transaction blocked: Insufficient available funds. Current ledger balance: ${self.balance:.2f}")

        # If all validation rules clear successfully, execute balancing ledger adjust
        self.balance -= amount
        print("\n" + "=" * 45)
        print(f"[TRANSACTION COMPLETE] Cash dispensed successfully!")
        print(f"Amount Withdrawn:  ${amount:.2f}")
        print(f"Updated Balance:   ${self.balance:.2f}")
        print("=" * 45)

def run_atm_session():
    # Instantiate simulation module registry
    atm = ATMSimulator(initial_balance=3500.0, true_pin="1234")
    print("=== AUTOMATED TELLER MACHINE (ATM) INTERFACE ===")

    # 1. Primary Authorization Phase Gate
    while True:
        try:
            atm.authenticate_user()
            break  # Break out of loop once credentials pass verification checks
        except InvalidPINError as e:
            print(f"\n[Security Alert] {e}\n")
            if atm.pin_attempts >= 3:
                print("Session terminated. Please contact your base branch bank administration.")
                return

    # 2. Main Transaction Dashboard Loop Phase
    while True:
        print("\n--- ACCOUNT DASHBOARD MENU ---")
        print("1. Check Balance")
        print("2. Withdraw Cash")
        print("3. Exit Terminal")
        
        choice = input("Select operation index (1/2/3): ").strip()
        print()

        try:
            if choice == '1':
                print(f"[Ledger Query] Your current active balance is: ${atm.balance:.2f}")
            elif choice == '2':
                atm.withdraw_cash()
            elif choice == '3':
                print("Retrieving card... Thank you for banking with us! System offline.")
                break
            else:
                print("[Menu Error] Invalid option mapping choice selected. Please select 1, 2, or 3.")
        
        # Intercept and gracefully capture custom application failure layers cleanly
        except InsufficientBalanceError as ib_err:
            print(f"\n[ATM Ledger Exception] {ib_err}")
        except InvalidWithdrawalAmountError as wa_err:
            print(f"\n[ATM Device Exception] {wa_err}")
        except Exception as gen_err:
            print(f"\n[System Core Exception] Unexpected runtime tracking fault: {gen_err}")

if __name__ == "__main__":
    run_atm_session()
```
