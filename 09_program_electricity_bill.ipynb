# Program 9: Global Variable – Electricity Bill

## Aim:
Write a Python program using a global variable to maintain the total electricity bill. Create functions to calculate units consumed and update the bill amount.

## Concepts Covered:
* Global variable scoping and the `global` keyword reference modifier
* Conditional tier-based slab calculation rules
* Encapsulated variable modification across isolated local namespaces


```python
# Primary global tracking variable maintaining the total bill statement across operations
TOTAL_BILL_AMOUNT = 0.0

def calculate_units_consumed(previous_reading, current_reading):
    """Computes total utility consumption units from meter state indices."""
    if current_reading < previous_reading:
        raise ValueError("Current reading cannot be lower than the previous reading.")
    return current_reading - previous_reading

def update_global_bill(units):
    """Calculates price matrix using progressive tier rates and updates global metric state."""
    global TOTAL_BILL_AMOUNT  # Instructs Python to link directly to the global context state
    
    # Tier-Based Progressive Slab Pricing Structure:
    # 1. First 100 units       -> $0.50 per unit
    # 2. Next 200 units (101-300) -> $0.75 per unit
    # 3. Above 300 units        -> $1.20 per unit
    
    calculated_charge = 0.0
    
    if units <= 100:
        calculated_charge = units * 0.50
    elif units <= 300:
        calculated_charge = (100 * 0.50) + ((units - 100) * 0.75)
    else:
        calculated_charge = (100 * 0.50) + (200 * 0.75) + ((units - 300) * 1.20)
        
    # Append the newly calculated charge structure onto our running global tracker variable
    TOTAL_BILL_AMOUNT += calculated_charge
    return calculated_charge

def run_billing_system():
    print("=== ELECTRICITY UTILITY BILLING INTERFACE ===")
    
    try:
        # Prompt user for mechanical meter state information
        prev = float(input("Enter previous meter reading (in kWh): ").strip())
        curr = float(input("Enter current meter reading (in kWh): ").strip())
        
        # Calculate metric values using function calls
        units = calculate_units_consumed(prev, curr)
        current_charge = update_global_bill(units)
        
        # Display individual transaction output logs
        print("\n" + "="*45)
        print(f"[SUCCESS] Consumption Record Extracted!")
        print(f"Total Units Consumed:    {units:.2f} kWh")
        print(f"Current Statement Cost:  ${current_charge:.2f}")
        print("-" * 45)
        print(f"TOTAL ACCOUNT BALANCE:   ${TOTAL_BILL_AMOUNT:.2f}")
        print("="*45 + "\n")
        
    except ValueError as e:
        print(f"\n[Calculation Error] Invalid numeric details provided: {e}\n")
    except Exception as e:
        print(f"\n[Unexpected Error] An issue occurred: {e}\n")

if __name__ == "__main__":
    # Simulate processing sequential billing sheets over multiple test iterations
    while True:
        run_billing_system()
        
        cont = input("Process another billing lifecycle statement? (y/n): ").strip().lower()
        print()
        if cont != 'y':
            print("System offline. Final aggregate summary dashboard logged.")
            print(f"Final Global Balance Stored: ${TOTAL_BILL_AMOUNT:.2f}")
            break
```
