# Program 10: Simple HTTP Server – Online Notice Board

## Aim:
Write a Python program to create a simple HTTP server that displays an online college notice board containing announcements and basic information.

## Concepts Covered:
* Web server infrastructure using Python's standard `http.server` library
* Dynamic response formatting with basic HTML structures
* Multi-threaded application routing using `threading` to keep the Jupyter interface responsive


```python
from http.server import BaseHTTPRequestHandler, HTTPServer
import threading
import time
import socket

class NoticeBoardHandler(BaseHTTPRequestHandler):
    """Custom request handler that outputs the HTML Notice Board."""
    
    def log_message(self, format, *args):
        # Override to suppress noisy server access log outputs in the notebook cell
        return

    def handle(self):
        """Override handle to gracefully trap connection drops (WinError 10053)."""
        try:
            super().handle()
        except (ConnectionAbortedError, ConnectionResetError):
            # Suppress noisy traceback console output if the browser closes the socket early
            pass

    def do_GET(self):
        try:
            self.send_response(200)
            self.send_header("Content-type", "text/html; charset=utf-8")
            self.end_headers()

            # HTML payload content for the Online College Notice Board
            html_content = """
            <!DOCTYPE html>
            <html>
            <head>
                <title>College Notice Board</title>
                <style>
                    body { font-family: Arial, sans-serif; margin: 40px; background-color: #f4f6f9; color: #333; }
                    .container { max-width: 800px; background: white; padding: 30px; border-radius: 8px; box-shadow: 0 4px 8px rgba(0,0,0,0.1); }
                    h1 { color: #003366; border-bottom: 3px solid #003366; padding-bottom: 10px; }
                    .notice { background-color: #fff9e6; border-left: 5px solid #ffcc00; padding: 15px; margin: 20px 0; border-radius: 4px; }
                    .notice h3 { margin-top: 0; color: #b38600; }
                    .notice-date { font-size: 0.85em; color: #666; font-style: italic; }
                    .footer { margin-top: 30px; font-size: 0.9em; color: #777; text-align: center; }
                </style>
            </head>
            <body>
                <div class="container">
                    <h1>🏛️ CAMPUS ONLINE NOTICE BOARD</h1>
                    <p>Welcome to the official digital bulletin system. Check recent announcements below:</p>
                    
                    <div class="notice">
                        <h3>📢 Semester Examination Schedule Out</h3>
                        <p>The final term examinations are scheduled to begin from next month. Please collect your hall tickets from the administration office before Friday.</p>
                        <p class="notice-date">Posted on: October 8, 2026</p>
                    </div>

                    <div class="notice">
                        <h3>🚀 Annual Tech Fest Registration</h3>
                        <p>Registrations for the upcoming hackathon and project exhibition tracks are now open. Visit the department coordinator to log your project squads.</p>
                        <p class="notice-date">Posted on: October 5, 2026</p>
                    </div>

                    <div class="notice" style="border-left-color: #cc0000; background-color: #ffe6e6;">
                        <h3 style="color: #cc0000;">⚠️ Network Maintenance Alert</h3>
                        <p>The campus Wi-Fi infrastructure will undergo routine system upgrades this Sunday from 02:00 AM to 06:00 AM. Expect temporary outages.</p>
                        <p class="notice-date">Posted on: October 2, 2026</p>
                    </div>

                    <div class="footer">
                        <p>Standard HTTP Server System Utility &bull; College Info Center</p>
                    </div>
                </div>
            </body>
            </html>
            """
            self.wfile.write(html_content.encode('utf-8'))
        except (ConnectionAbortedError, ConnectionResetError):
            # Catch errors during transmission stream writes
            pass

def run_server(start_port=8085):
    """Binds and runs the HTTP server platform framework, scanning for an open port if needed."""
    port = start_port
    while port < start_port + 20:
        try:
            server_address = ('127.0.0.1', port)
            httpd = HTTPServer(server_address, NoticeBoardHandler)
            print(f"[Server] System operational. Notice Board live at http://127.0.0.1:{port}")
            print("Click the link above to view your server webpage safely.\n")
            httpd.serve_forever()
            return
        except OSError:
            port += 1
            
    print("[Server Error] Failed to bind to any available ports in the specified range.")

if __name__ == "__main__":
    # Start on a higher port number to avoid standard permission filters
    START_PORT = 8085
    
    server_thread = threading.Thread(target=run_server, args=(START_PORT,), daemon=True)
    server_thread.start()
    
    time.sleep(0.5)
```
