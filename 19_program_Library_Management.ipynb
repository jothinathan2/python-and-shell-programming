# Program 19: User-Defined Module – Library Management

## Aim:
Create a user-defined Python module containing classes and functions for managing library books. Import the module into another program and perform book issue and return operations.

## Concepts Covered:
* Modular Programming using Jupyter file writing magics (`%%writefile`)
* Object-Oriented Programming (Classes, Methods, Object Instantiation)
* Namespace manipulation and explicit module importing semantics


```python
# =====================================================================
# PART 1: WRITING THE USER-DEFINED MODULE TO DISK
# This special Jupyter magic line creates a real python module file locally.
# =====================================================================
import sys

module_code = """
class Book:
    \"\"\"Class representing an individual book asset.\"\"\"
    def __init__(self, book_id, title, author):
        self.book_id = book_id
        self.title = title
        self.author = author
        self.is_issued = False  # Track availability state

class Library:
    \"\"\"Class managing a collection of books.\"\"\"
    def __init__(self):
        self.books = {}  # Master registry using Book ID as the key

    def add_book(self, book_id, title, author):
        \"\"\"Registers a new book object into the system.\"\"\"
        new_book = Book(book_id, title, author)
        self.books[book_id] = new_book
        return f"Added: '{title}' by {author} [ID: {book_id}]"

    def issue_book(self, book_id):
        \"\"\"Changes book state to issued if available.\"\"\"
        if book_id not in self.books:
            return f"[Error] Book ID {book_id} does not exist in this library."
        
        book = self.books[book_id]
        if book.is_issued:
            return f"[Transaction Blocked] '{book.title}' is already issued to another user."
        
        book.is_issued = True
        return f"[Success] '{book.title}' has been successfully issued."

    def return_book(self, book_id):
        \"\"\"Reverts book state back to available.\"\"\"
        if book_id not in self.books:
            return f"[Error] Book ID {book_id} does not exist in this library."
        
        book = self.books[book_id]
        if not book.is_issued:
            return f"[Transaction Info] '{book.title}' was already inside the library inventory."
        
        book.is_issued = False
        return f"[Success] '{book.title}' has been returned to stock."

    def display_all_books(self):
        \"\"\"Returns tabular summary data of all indexed books.\"\"\"
        if not self.books:
            return "The library inventory is completely empty."
            
        output = "\\n" + "="*60 + "\\n"
        output += f"{'ID':<10}{'Title':<30}{'Status':<15}\\n"
        output += "="*60 + "\\n"
        for b_id, book in self.books.items():
            status = "Issued Out" if book.is_issued else "Available"
            output += f"{b_id:<10}{book.title:<30}{status:<15}\\n"
        output += "="*60
        return output
"""

# Write the string out into a local library file
with open("library_management.py", "w", encoding="utf-8") as f:
    f.write(module_code.strip())
print("[System Setup] 'library_management.py' module file created successfully in your working directory.")

# =====================================================================
# PART 2: IMPORTING AND RUNNING THE GENERATED USER MODULE
# =====================================================================
# Import the custom module we just generated on disk
import library_management

def run_library_system():
    print("\n=== SYSTEM INTERFACE: RUNNING IMPORTED LIBRARY MODULE ===")
    
    # Instantiate the Library core system from the user-defined module
    my_library = library_management.Library()
    
    # 1. Populate the database via system rules
    my_library.add_book("B101", "Python Programming Core", "Guido van Rossum")
    my_library.add_book("B102", "Computer Networks", "Andrew Tanenbaum")
    my_library.add_book("B103", "Introduction to Algorithms", "Thomas Cormen")
    
    # Display starting inventory state
    print("\n--- Initial Catalog State ---")
    print(my_library.display_all_books())
    
    # 2. Simulate standard operational loops interactively
    while True:
        print("\n--- OPERATIONS PANEL ---")
        print("1. Issue a Book")
        print("2. Return a Book")
        print("3. View Current Book Catalog")
        print("4. Exit System")
        
        choice = input("Select operation choice (1-4): ").strip()
        print()
        
        if choice == '1':
            target_id = input("Enter Book ID to issue (e.g., B101, B102): ").strip()
            # Call function directly hosted on the user module class
            result = my_library.issue_book(target_id)
            print(result)
            
        elif choice == '2':
            target_id = input("Enter Book ID to return (e.g., B101): ").strip()
            result = my_library.return_book(target_id)
            print(result)
            
        elif choice == '3':
            print(my_library.display_all_books())
            
        elif choice == '4':
            print("Closing application pipeline. Library registry saved securely.")
            break
        else:
            print("Invalid input! Please select index from 1 to 4.")

if __name__ == "__main__":
    run_library_system()
```
