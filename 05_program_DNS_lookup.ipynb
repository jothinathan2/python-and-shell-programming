# Program 5: DNS Lookup – Network Utility

## Aim:
Write a Python program to accept a domain name from the user and retrieve its corresponding IP address using DNS lookup.

## Concepts Covered:
* Network name resolution using `socket.gethostbyname()`
* Input parsing and URL cleaning strings
* Graceful handling of invalid domains via `socket.gaierror`


```python
import socket
from urllib.parse import urlparse

def perform_dns_lookup():
    print("=== DNS LOOKUP NETWORK UTILITY ===")
    
    # Accept domain name input from the user
    user_input = input("Enter a domain name to lookup (e.g., google.com or ://github.com): ").strip()
    
    # Basic sanitization: if user copies a full URL, strip out http://, https://, and paths
    domain = user_input
    if "://" in domain:
        parsed_url = urlparse(domain)
        domain = parsed_url.netloc
    elif "/" in domain:
        domain = domain.split('/')[0]
        
    print(f"\n[Processing] Querying DNS servers for domain: '{domain}'...")
    
    try:
        # Perform the standard DNS lookup to fetch the IPv4 address
        ip_address = socket.gethostbyname(domain)
        
        print("\n" + "="*45)
        print(f"[SUCCESS] DNS Resolution Complete!")
        print(f"Domain Name:  {domain}")
        print(f"IP Address:   {ip_address}")
        print("="*45 + "\n")
        
    except socket.gaierror:
        # Handles Address Info errors (e.g., if the host name cannot be found or is invalid)
        print("\n" + "="*45)
        print(f"[DNS Error] Could not resolve host '{domain}'.")
        print("Please check the domain spelling or your network connection.")
        print("="*45 + "\n")
        
    except Exception as e:
        print(f"\n[Unexpected Error] An issue occurred: {e}")

if __name__ == "__main__":
    perform_dns_lookup()
```
