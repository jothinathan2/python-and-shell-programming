# Program 16: Multi-Client Socket Server – Chat Room

## Aim:
Write a Python program to create a socket-based server that accepts multiple clients and broadcasts a simple message to connected clients.

## Concepts Covered:
* Concurrent request handling via multi-threaded server routing
* Dynamic thread tracking mapping configurations (`list.append` / `list.remove`)
* Direct data broadcast replication loops over connected socket arrays


```python
import socket
import threading
import time

def broadcast_message(message_string, active_clients_list):
    """Iterates through active connections to broadcast raw data streams."""
    encoded_payload = message_string.encode('utf-8')
    for client_socket in active_clients_list[:]:
        try:
            client_socket.sendall(encoded_payload)
        except Exception:
            # Handle sudden socket drops by dropping the socket handle
            if client_socket in active_clients_list:
                active_clients_list.remove(client_socket)

def run_chat_server(host, port, client_connections_registry):
    """Asynchronous socket server managing dynamic client loops."""
    try:
        with socket.socket(socket.AF_INET, socket.SOCK_STREAM) as server_sock:
            server_sock.setsockopt(socket.SOL_SOCKET, socket.SO_REUSEADDR, 1)
            server_sock.bind((host, port))
            server_sock.listen(5)
            print(f"[Broadcast Server] System live on {host}:{port}. Awaiting nodes...\n")
            
            # Continuous processing listener engine loop
            while True:
                client_sock, client_addr = server_sock.accept()
                print(f"[Server Connection] New node linked from: {client_addr}")
                
                # Append client context to tracking registry lists
                client_connections_registry.append(client_sock)
                
                # Instantly broadcast a confirmation footprint down to all endpoints
                broadcast_message(f"\n[System Alert] Node {client_addr} entered the room.", client_connections_registry)
                
    except Exception as e:
        pass

def simulate_client_connection(host, port, client_name):
    """Simulates an isolated client terminal instance mapping context."""
    try:
        with socket.socket(socket.AF_INET, socket.SOCK_STREAM) as client_sock:
            client_sock.connect((host, port))
            print(f" -> [{client_name}] Established path. Listening for server packets...")
            
            # Read greeting confirmation packet from stream pipeline
            server_packet = client_sock.recv(1024).decode('utf-8')
            print(f" -> [{client_name} Received Log]: {server_packet.strip()}")
    except Exception as e:
        print(f" -> [{client_name} Fault]: Connection rejected. {e}")

if __name__ == "__main__":
    HOST_IP = '127.0.0.1'
    PORT_NO = 65450
    active_connections = []
    
    print("=== MULTI-CLIENT SOCKET BROADCAST LAB ===")
    
    # 1. Fire up Server instance thread context securely as a background process
    server_node = threading.Thread(target=run_chat_server, args=(HOST_IP, PORT_NO, active_connections), daemon=True)
    server_node.start()
    
    # 2. Allow a brief split-second window for port allocation setup bounds
    time.sleep(0.4)
    
    # 3. Simulate sequential client context triggers hitting the shared interface channel
    print("\n[Simulation Run] Dispatching Client Node 1...")
    c1 = threading.Thread(target=simulate_client_connection, args=(HOST_IP, PORT_NO, "Client_Alpha"))
    c1.start()
    time.sleep(0.3)
    
    print("\n[Simulation Run] Dispatching Client Node 2...")
    c2 = threading.Thread(target=simulate_client_connection, args=(HOST_IP, PORT_NO, "Client_Beta"))
    c2.start()
    time.sleep(0.5)
    
    # 4. Issue a broad master notification broadcast right from the root execution engine context
    print("\n[Simulation Run] Server issues an enterprise broadcast notice:")
    broadcast_message("[GLOBAL NOTICE] This lab session will safely expire in 10 seconds.", active_connections)
    time.sleep(0.5)
```
