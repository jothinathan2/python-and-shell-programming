# Program 7: List Operations – Shopping Cart

## Aim:
Write a Python program to maintain a shopping cart using a list. Add, remove and display products using list operations and slicing.

## Concepts Covered:
* Dynamic array manipulation via Python lists (`append`, `remove`)
* Element parsing and data boundaries using list slicing (`[start:stop]`)
* Input routing validation checks for sequential indices


```python
def add_item(cart):
    """Appends a new product into the shopping cart array."""
    print("--- Add Product ---")
    item = input("Enter the name of the product to add: ").strip()
    if item:
        cart.append(item)
        print(f"[Success] '{item}' added to your shopping cart.\n")
    else:
        print("[Error] Product name cannot be blank.\n")

def remove_item(cart):
    """Removes a specific product match from the shopping cart array."""
    print("--- Remove Product ---")
    if not cart:
        print("Your cart is empty. Nothing to remove.\n")
        return
        
    item = input("Enter the exact name of the product to remove: ").strip()
    if item in cart:
        cart.remove(item)
        print(f"[Success] '{item}' has been removed from your cart.\n")
    else:
        print(f"[Error] '{item}' was not found in your shopping cart.\n")

def display_cart(cart):
    """Displays all current cart items, including demonstration of slicing operations."""
    print("--- Current Shopping Cart Summary ---")
    if not cart:
        print("Your shopping cart is currently empty.\n")
        return
        
    print(f"Total Items in Cart: {len(cart)}")
    print("Full Cart List:", cart)
    
    print("\n" + "="*45)
    print("Index \t Product Name")
    print("="*45)
    for index, item in enumerate(cart):
        print(f"[{index}] \t {item}")
    print("="*45)
    
    # Showcase standard slicing operations
    if len(cart) >= 2:
        print(f"\n[Slicing Demo] First two items: {cart[:2]}")
        print(f"[Slicing Demo] Last item in cart: {cart[-1:]}")
    print()

# Main execution loop
if __name__ == "__main__":
    # Primary application list acting as the shopping cart repository
    shopping_cart = []
    
    while True:
        print("=== SHOPPING CART INTERFACE ===")
        print("1. Add Product to Cart")
        print("2. Remove Product from Cart")
        print("3. Display Cart Summary & Slicing Operations")
        print("4. Exit")
        
        choice = input("Enter your choice (1/2/3/4): ").strip()
        print()
        
        if choice == '1':
            add_item(shopping_cart)
        elif choice == '2':
            remove_item(shopping_cart)
        elif choice == '3':
            display_cart(shopping_cart)
        elif choice == '4':
            print("Exiting Shopping Cart Application. Goodbye!")
            break
        else:
            print("Invalid choice! Please select an option from 1 to 4.\n")
                                                                                                                                                                                        
```
