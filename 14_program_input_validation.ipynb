# Program 14: Input Validation – Secure Login

## Aim:
Write a Python program for a login system that validates username and password input and displays appropriate error messages for invalid input.

## Concepts Covered:
* Input validation rules (length limits, character safety boundaries)
* Custom application error flagging arrays
* Data sanitization and pattern constraint verification


```python
def validate_credentials(username, password):
    """Validates login credentials against enterprise complexity and safety rules."""
    errors = []
    
    # ------------------ Username Validation Layers ------------------
    # Rule 1: Length enforcement checks
    if len(username) < 4 or len(username) > 16:
        errors.append("Username must be between 4 and 16 characters long.")
        
    # Rule 2: Allowed character pattern verification (alphanumeric and underscores only)
    # This acts as a primary defense boundary against primitive SQL injection patterns
    if not all(char.isalnum() or char == '_' for char in username):
        errors.append("Username can only contain alphanumeric characters and underscores (_).")
        
    # ------------------ Password Validation Layers ------------------
    # Rule 3: Length enforcement checks
    if len(password) < 8:
        errors.append("Password must be at least 8 characters long.")
        
    # Rule 4: Structural character variation verification
    has_upper = any(char.isupper() for char in password)
    has_lower = any(char.islower() for char in password)
    has_digit = any(char.isdigit() for char in password)
    has_special = any(not char.isalnum() for char in password)
    
    if not (has_upper and has_lower and has_digit and has_special):
        errors.append("Password must include uppercase letters, lowercase letters, numbers, and special characters.")
        
    # Rule 5: Blocklist validation against common weak credential phrases
    weak_passwords_blocklist = ["password123", "admin123", "welcome123", "qwertyuiop"]
    if password.lower() in weak_passwords_blocklist:
        errors.append("Entered password is on the global weak-credential blocklist.")
        
    return errors

def run_login_portal():
    print("=== SECURE ENTERPRISE LOGIN PORTAL ===")
    
    # Capture user authentication strings safely, stripping surrounding whitespace
    username_input = input("Enter Username: ").strip()
    password_input = input("Enter Password: ").strip()
    
    print("\n[Processing] Evaluating credentials against security profiles...")
    
    # Run structural verification validation logic
    validation_faults = validate_credentials(username_input, password_input)
    
    print("\n" + "=" * 55)
    if not validation_faults:
        print("[ACCESS GRANTED] Inputs pass security validation profiles.")
        print(f"Welcome back, corporate profile session: '{username_input}'.")
    else:
        print("[ACCESS DENIED] Input Validation Failures Found!")
        print("-" * 55)
        # Enumerate and log each broken logic statement cleanly for the user interface
        for index, fault in enumerate(validation_faults, 1):
            print(f" -> Error {index}: {fault}")
    print("=" * 55 + "\n")

if __name__ == "__main__":
    run_login_portal()
```
